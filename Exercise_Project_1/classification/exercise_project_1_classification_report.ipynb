## Data Dictionary

**Source:** UCI Heart Disease (Cleveland) dataset, 303 patients, 14 columns.
**Target:** `target` (1 = heart disease present, 0 = not present).

### Patient information
| Column | Description | Values |
|---|---|---|
| `age` | Age in years | 29 to 77 |
| `sex` | Sex | 1 = male, 0 = female |

### Symptoms and basic measurements
| Column | Description | Values |
|---|---|---|
| `cp` | Chest pain type | 0 = typical angina, 1 = atypical angina, 2 = non-anginal pain, 3 = asymptomatic |
| `trestbps` | Resting blood pressure (mm Hg) | 94 to 200 |
| `chol` | Serum cholesterol (mg/dl) | 126 to 564 |
| `fbs` | Fasting blood sugar > 120 mg/dl | 1 = yes, 0 = no |
| `restecg` | Resting ECG result | 0 = normal, 1 = ST-T abnormality, 2 = left ventricular hypertrophy |

### Exercise stress test results
| Column | Description | Plain-English meaning |
|---|---|---|
| `thalach` | Maximum heart rate achieved (bpm) | How high the heart rate got during exercise |
| `exang` | Exercise-induced angina (1 = yes, 0 = no) | Did the patient feel chest pain during exercise? |
| `oldpeak` | ST depression induced by exercise relative to rest | How much the ECG line dipped under exercise (higher = more concern) |
| `slope` | Slope of the peak exercise ST segment (0 = up, 1 = flat, 2 = down) | Direction the ECG line trended at peak effort |

### Other tests
| Column | Description | Plain-English meaning |
|---|---|---|
| `ca` | Number of major vessels (0 to 3) colored by fluoroscopy | How many main heart vessels look blocked |
| `thal` | Thalassemia test result | 1 = fixed defect, 2 = normal, 3 = reversible defect |
